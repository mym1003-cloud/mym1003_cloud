# 수식 계산 & 그래프
SymPy + Matplotlib 활용

```python
%config InlineBackend.figure_format = "retina"
import numpy as np
import matplotlib.pyplot as plt
from sympy import *

plt.rcParams['font.family'] = 'Malgun Gothic'  # 한글 폰트
plt.rcParams['axes.unicode_minus'] = False
x = symbols('x')
print('준비 완료!')
```

## 1. 함수 그래프 + 기울기(미분)

```python
# 함수 정의
f = x**3 - 2*x**2 + x + 1
f_prime = diff(f, x)  # 미분 (기울기)

print('f(x) =', f)
print("f'(x) =", f_prime)

# 그래프
x_vals = np.linspace(-1, 3, 300)
f_vals = [float(f.subs(x, v)) for v in x_vals]
fp_vals = [float(f_prime.subs(x, v)) for v in x_vals]

plt.figure(figsize=(10, 5))
plt.plot(x_vals, f_vals, label='f(x)', linewidth=2)
plt.plot(x_vals, fp_vals, label="f'(x) 기울기", linestyle='--', linewidth=2)
plt.axhline(0, color='black', linewidth=0.5)
plt.legend()
plt.title('함수와 기울기(미분)')
plt.grid(True)
plt.show()
```

## 2. 성장 함수 비교

```python
x_vals = np.linspace(0.1, 5, 300)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, np.log(x_vals),      label='log(x)')
plt.plot(x_vals, x_vals,              label='x (선형)')
plt.plot(x_vals, x_vals**2,           label='x² (이차)')
plt.plot(x_vals, np.exp(x_vals),      label='eˣ (지수)')
plt.plot(x_vals, x_vals*np.log(x_vals), label='x·log(x)')

plt.ylim(-2, 30)
plt.legend()
plt.title('성장 함수 비교')
plt.xlabel('x')
plt.grid(True)
plt.show()
```

## 3. 수식 기호 계산 (적분, 극한, 급수)

```python
x, n = symbols('x n')

# 적분
f = sin(x) * exp(-x)
print('적분:', integrate(f, x))

# 극한
print('극한 (x→0):', limit(sin(x)/x, x, 0))

# 테일러 급수
print('sin(x) 테일러:', series(sin(x), x, 0, 7))
```
