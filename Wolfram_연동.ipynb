# Wolfram Engine 연동

설치된 Wolfram Engine 14.3을 Python에서 직접 사용합니다.

```python
from wolframclient.evaluation import WolframLanguageSession
from wolframclient.language import wl, wlexpr

# Wolfram Engine 경로
WOLFRAM_PATH = r'C:\Program Files\Wolfram Research\Wolfram Engine\14.3\WolframKernel.exe'

session = WolframLanguageSession(WOLFRAM_PATH)
print('Wolfram Engine 연결 성공!')
```

```python
# 수식 계산
result = session.evaluate(wlexpr('Integrate[Sin[x]*Exp[-x], x]'))
print('적분 결과:', result)

result2 = session.evaluate(wlexpr('D[x^3 - 2x^2 + x, x]'))
print('미분 결과:', result2)

result3 = session.evaluate(wlexpr('Limit[Sin[x]/x, x->0]'))
print('극한값:', result3)

result4 = session.evaluate(wlexpr('NSolve[x^2 - 5x + 6 == 0, x]'))
print('방정식 해:', result4)
```

```python
# Wolfram으로 그래프 생성 후 matplotlib으로 표시
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'

# Wolfram에서 수치 데이터 계산
data = session.evaluate(wlexpr('Table[{x, Sin[x]*Exp[-x]}, {x, 0, 10, 0.1}]'))

xs = [float(d[0]) for d in data]
ys = [float(d[1]) for d in data]

plt.figure(figsize=(10, 5))
plt.plot(xs, ys, lw=2, color='royalblue')
plt.title('sin(x)·e⁻ˣ  (Wolfram Engine 계산)', fontsize=14)
plt.grid(True)
plt.show()
```

```python
# 세션 종료
session.terminate()
print('Wolfram Engine 종료')
```
