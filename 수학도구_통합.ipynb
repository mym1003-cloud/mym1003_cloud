## 1. Desmos — 브라우저에서 바로 열기

```python
import webbrowser
from IPython.display import IFrame, display, HTML

# Desmos 그래프를 VS Code 내부에서 표시
display(HTML('<a href="https://www.desmos.com/calculator" target="_blank" style="font-size:18px; padding:10px; background:#2d70b3; color:white; border-radius:5px; text-decoration:none;">🔗 Desmos 열기</a>'))

# 특정 함수 미리 입력된 Desmos 링크 생성
def desmos_link(expression='x^2'):
    import urllib.parse
    encoded = urllib.parse.quote(expression)
    url = f'https://www.desmos.com/calculator?expression={encoded}'
    display(HTML(f'<a href="{url}" target="_blank">Desmos에서 {expression} 보기</a>'))

desmos_link('x^3 - 2x^2 + x')
desmos_link('e^x')
desmos_link('\\sin(x)')
```

## 2. Wolfram Alpha — 수식 풀이

```python
from IPython.display import display, HTML

def wolfram_link(query):
    import urllib.parse
    encoded = urllib.parse.quote(query)
    url = f'https://www.wolframalpha.com/input?i={encoded}'
    display(HTML(f'<a href="{url}" target="_blank" style="display:inline-block; margin:4px; padding:8px 12px; background:#dd1100; color:white; border-radius:4px; text-decoration:none;">WA: {query}</a>'))

# 클릭하면 Wolfram Alpha에서 바로 계산
wolfram_link('integrate sin(x)*e^(-x) dx')
wolfram_link('derivative of x^3 - 2x^2 + x')
wolfram_link('limit sin(x)/x as x->0')
wolfram_link('solve x^2 - 5x + 6 = 0')
wolfram_link('taylor series of e^x')
```

## 3. Python SymPy — 코드로 직접 계산

```python
from sympy import *
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

x = symbols('x')

# ---- 원하는 함수를 여기서 바꾸세요 ----
f = x**3 - 2*x**2 + x
# --------------------------------------

f_prime  = diff(f, x)       # 1차 미분 (기울기)
f_prime2 = diff(f, x, 2)    # 2차 미분
f_integral = integrate(f, x) # 적분

print('f(x)      =', f)
print("f'(x)     =", f_prime,  ' ← 기울기')
print("f''(x)    =", f_prime2, ' ← 볼록/오목')
print('∫f(x)dx   =', f_integral)

# 그래프
xv = np.linspace(-1, 3, 400)
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(xv, [float(f.subs(x,v)) for v in xv],       label='f(x)',      lw=2)
ax.plot(xv, [float(f_prime.subs(x,v)) for v in xv], label="f'(x) 기울기", lw=2, ls='--')
ax.axhline(0, color='black', lw=0.7)
ax.legend(fontsize=12)
ax.set_title('함수 + 기울기', fontsize=14)
ax.grid(True)
plt.show()
```

## 4. 성장 함수 비교

```python
xv = np.linspace(0.1, 5, 400)
functions = {
    'log(x)':   np.log(xv),
    'x (선형)': xv,
    'x² (이차)': xv**2,
    'x³ (삼차)': xv**3,
    'eˣ (지수)': np.exp(xv),
    'x·log(x)': xv * np.log(xv),
}

plt.figure(figsize=(11, 6))
for name, vals in functions.items():
    plt.plot(xv, np.clip(vals, -5, 50), label=name, lw=2)

plt.legend(fontsize=11)
plt.title('성장 함수 비교', fontsize=14)
plt.xlabel('x')
plt.ylim(-5, 50)
plt.grid(True)
plt.show()
```
