# GeoGebra 연동

GeoGebra는 VS Code 확장이 없어 **웹 앱**으로 연동합니다.

- 그래프: https://www.geogebra.org/graphing
- 기하학: https://www.geogebra.org/geometry
- 3D: https://www.geogebra.org/3d
- 계산기: https://www.geogebra.org/calculator

```python
from IPython.display import display, HTML

tools = [
    ('그래프 (2D)', 'https://www.geogebra.org/graphing', '#4CAF50'),
    ('기하학',     'https://www.geogebra.org/geometry', '#2196F3'),
    ('3D 그래프',  'https://www.geogebra.org/3d',       '#9C27B0'),
    ('계산기',     'https://www.geogebra.org/calculator','#FF5722'),
    ('CAS 수식',   'https://www.geogebra.org/cas',      '#795548'),
]

html = '<h3>GeoGebra 도구 선택</h3>'
for name, url, color in tools:
    html += f'<a href="{url}" target="_blank" style="display:inline-block; margin:6px; padding:12px 20px; background:{color}; color:white; border-radius:6px; text-decoration:none; font-size:15px;">📐 {name}</a>'
display(HTML(html))
```

```python
# GeoGebra에 함수 미리 입력해서 열기
import urllib.parse
from IPython.display import display, HTML

def geogebra_graph(expression):
    encoded = urllib.parse.quote(expression)
    url = f'https://www.geogebra.org/graphing?lang=ko'
    display(HTML(f'<a href="{url}" target="_blank" style="padding:8px 14px; background:#4CAF50; color:white; border-radius:4px; text-decoration:none;">GeoGebra에서 열기: {expression}</a><br><br>'))

geogebra_graph('f(x) = x^3 - 2x^2 + x')
geogebra_graph('g(x) = sin(x) * e^(-x)')
geogebra_graph('h(x) = ln(x)')
```
